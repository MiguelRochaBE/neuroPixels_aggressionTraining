```python
import re, cv2, numpy as np, matplotlib.pyplot as plt
from matplotlib.widgets import Slider, Button, TextBox
from pathlib import Path

# ================== CONFIG ==================
VIDEO_DIRS = [Path(r"C:\Users\stagk\OneDrive\Desktop\trainingAggression_MR\trainingSessions\unifiedDataset_topview_DLC_singleAnimal\analyzedVideos_Stefanos_RIdataset_M2_DLC_singleAnimal")]
VIDEO_EXTS = (".mp4", ".avi", ".mov", ".mkv")
LAT_DIMS   = (0, 1)
VIDEO_OVERRIDE = {}
VIDEO_REGION  = (0.02, 0.10, 0.70, 0.86)
PLAY_INTERVAL = 30                              # ms between ticks (lower = faster)
# ============================================

def parse_key(k):
    m = re.search(r"mouse(\d+)_Day0*(\d+)", str(k))
    return (m.group(1), m.group(2)) if m else (None, None)

avail = [k for k in resultsDataset["video"] if k in results]
by_mouse = {}
for k in avail:
    mo, dy = parse_key(k)
    if mo: by_mouse.setdefault(mo, {})[dy] = k
mice = sorted(by_mouse, key=lambda m: int(m))
def days_for(m): return sorted(by_mouse[m], key=lambda d: int(d))

def find_video_file(key):
    if key in VIDEO_OVERRIDE: return Path(VIDEO_OVERRIDE[key])
    mo, dy = parse_key(key)
    if mo is None: return None
    for base in VIDEO_DIRS:
        if not base.exists(): continue
        cands = [p for p in base.rglob("*") if p.suffix.lower() in VIDEO_EXTS and f"mouse{mo}" in p.stem]
        for p in cands:
            m2 = re.search(r"[Dd]ay0*(\d+)", p.stem)
            if m2 and m2.group(1) == dy: return p
        if cands: return cands[0]
    return None

S = {}
def load_session(key):
    r = results[key]
    cen = np.asarray(r["centroid"]); head = np.asarray(r["heading"])
    lat = np.asarray(r["latent_state"]); syl = np.asarray(r["syllable"]).astype(int)
    L2  = lat[:, list(LAT_DIMS)]
    if S.get("cap") is not None: S["cap"].release()
    vf = find_video_file(key); cap = cv2.VideoCapture(str(vf)) if vf and vf.exists() else None
    fw = fh = None; nvid = len(cen)
    if cap:
        nvid = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) or len(cen)
        ok, fr0 = cap.read()
        if ok: fh, fw = fr0.shape[:2]
        cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
    step = max(1, round(nvid / len(cen)))
    return dict(cen=cen, head=head, L2=L2, syl=syl, cap=cap, step=step,
                n=len(cen), vf=vf, fw=fw, fh=fh, bg=None)

def get_frame(cap, vidx):
    vidx = int(vidx); pos = int(cap.get(cv2.CAP_PROP_POS_FRAMES))
    if vidx < pos or vidx - pos > 8:
        cap.set(cv2.CAP_PROP_POS_FRAMES, vidx)          # random access -> seek
    else:
        for _ in range(vidx - pos): cap.grab()          # short hop -> cheap skip
    ok, fr = cap.read()
    return cv2.cvtColor(fr, cv2.COLOR_BGR2RGB) if ok else None

# ---------- figure + axes ----------
fig = plt.figure(figsize=(16, 9)); fig.canvas.manager.set_window_title("KPMS session viewer")
ax_v = fig.add_axes(list(VIDEO_REGION))
ax_t = fig.add_axes([0.75, 0.52, 0.23, 0.34])
ax_l = fig.add_axes([0.75, 0.10, 0.23, 0.34])
ax_v.set_xticks([]); ax_v.set_yticks([])
im  = ax_v.imshow(np.zeros((10, 10, 3), np.uint8), aspect="auto", animated=True); ax_v.set_title("video")
txt = ax_v.text(0.97, 0.97, "", transform=ax_v.transAxes, ha="right", va="top", animated=True,
                fontsize=18, color="white", weight="bold", bbox=dict(boxstyle="round", fc="black", alpha=0.6))
bg_t,   = ax_t.plot([], [], "-", color="0.8", lw=0.6)
trail,  = ax_t.plot([], [], "-", color="tab:blue", lw=1.5, animated=True)
cur,    = ax_t.plot([], [], "o", color="red", ms=7, animated=True)
q       = ax_t.quiver([0], [0], [0], [0], color="red", scale=12, width=0.012, animated=True)
ax_t.set_title("centroid + heading"); ax_t.invert_yaxis()
bg_l,   = ax_l.plot([], [], "-", color="0.8", lw=0.6)
ltrail, = ax_l.plot([], [], "-", color="tab:green", lw=1.5, animated=True)
lcur,   = ax_l.plot([], [], "o", color="red", ms=7, animated=True)
ax_l.set_title(f"latent dims {LAT_DIMS[0]} vs {LAT_DIMS[1]}")
DYN = [im, txt, trail, cur, q, ltrail, lcur]

def fit_video_axes():
    if not S or not S.get("fw"):
        ax_v.set_position(list(VIDEO_REGION)); return
    w, h = S["fw"], S["fh"]; figW, figH = fig.get_size_inches()
    rx, ry, rw, rh = VIDEO_REGION
    bh = rh; bw = (w / h) * (figH / figW) * bh
    if bw > rw: bw = rw; bh = (h / w) * (figW / figH) * bw
    ax_v.set_position([rx + (rw - bw) / 2, ry + (rh - bh) / 2, bw, bh])

def refresh_limits():
    c = S["cen"]
    ax_t.set_xlim(np.nanmin(c[:,0]), np.nanmax(c[:,0])); ax_t.set_ylim(np.nanmax(c[:,1]), np.nanmin(c[:,1]))
    bg_t.set_data(c[:,0], c[:,1])
    L = S["L2"]
    ax_l.set_xlim(np.nanmin(L[:,0]), np.nanmax(L[:,0])); ax_l.set_ylim(np.nanmin(L[:,1]), np.nanmax(L[:,1]))
    bg_l.set_data(L[:,0], L[:,1])
    if S["fw"]:
        im.set_extent([0, S["fw"], S["fh"], 0]); ax_v.set_xlim(0, S["fw"]); ax_v.set_ylim(S["fh"], 0)

def prime_bg():
    for a in DYN: a.set_visible(False)
    fig.canvas.draw(); S["bg"] = fig.canvas.copy_from_bbox(fig.bbox)
    for a in DYN: a.set_visible(True)

def update(t):
    if not S: return
    t = int(t); w = int(s_trail.val); a = max(0, t - w)
    if S["cap"] is not None:
        fr = get_frame(S["cap"], t * S["step"])
        if fr is not None: im.set_data(fr)
    txt.set_text(f"syllable {S['syl'][t]}")
    c = S["cen"]; trail.set_data(c[a:t+1,0], c[a:t+1,1]); cur.set_data([c[t,0]], [c[t,1]])
    q.set_offsets(np.array([c[t]])); q.set_UVC(np.cos(S["head"][t]), np.sin(S["head"][t]))
    L = S["L2"]; ltrail.set_data(L[a:t+1,0], L[a:t+1,1]); lcur.set_data([L[t,0]], [L[t,1]])
    if S.get("bg") is None: prime_bg()
    fig.canvas.restore_region(S["bg"])
    for art in DYN: art.axes.draw_artist(art)
    fig.canvas.blit(fig.bbox); fig.canvas.flush_events()

# ---------- controls ----------
ax_pm = fig.add_axes([0.03, 0.02, 0.05, 0.045]); b_pm = Button(ax_pm, "◀ M")
ax_nm = fig.add_axes([0.09, 0.02, 0.05, 0.045]); b_nm = Button(ax_nm, "M ▶")
ax_pd = fig.add_axes([0.16, 0.02, 0.05, 0.045]); b_pd = Button(ax_pd, "◀ D")
ax_nd = fig.add_axes([0.22, 0.02, 0.05, 0.045]); b_nd = Button(ax_nd, "D ▶")
ax_fr = fig.add_axes([0.33, 0.058, 0.40, 0.022]); s_frame = Slider(ax_fr, "frame", 0, 1, valinit=0, valstep=1)
ax_tr = fig.add_axes([0.33, 0.022, 0.40, 0.022]); s_trail = Slider(ax_tr, "trail", 5, 400, valinit=60, valstep=5)
ax_sp = fig.add_axes([0.80, 0.058, 0.15, 0.022]); s_speed = Slider(ax_sp, "speed", 1, 30, valinit=2, valstep=1)
ax_pl = fig.add_axes([0.80, 0.012, 0.08, 0.038]); b_play = Button(ax_pl, "Play")
ax_sr = fig.add_axes([0.80, 0.93, 0.18, 0.045]); tb = TextBox(ax_sr, "mouse ", initial="")
s_frame.drawon = False

state = {"mi": 0, "di": 0}
def cur_key():
    m = mice[state["mi"]]; d = days_for(m)[state["di"]]; return by_mouse[m][d]

def reload_rec():
    global S
    key = cur_key(); S = load_session(key)
    s_frame.valmax = max(1, S["n"] - 1); s_frame.ax.set_xlim(0, s_frame.valmax)
    refresh_limits(); fit_video_axes()
    res = f"{S['fw']}×{S['fh']}" if S["fw"] else "?"
    vlabel = S["vf"].name if S["vf"] else "video not found (trajectories only)"
    fig.suptitle(f"mouse {mice[state['mi']]}  {cur_key()}  |  {S['n']} frames, step×{S['step']}, ROI {res}", fontsize=10)
    S["bg"] = None
    s_frame.eventson = False; s_frame.set_val(0); s_frame.eventson = True
    update(0)

def nav_mouse(d): state["mi"] = (state["mi"] + d) % len(mice); state["di"] = 0; reload_rec()
def nav_day(d):   n = len(days_for(mice[state["mi"]])); state["di"] = (state["di"] + d) % n; reload_rec()
b_pm.on_clicked(lambda e: nav_mouse(-1)); b_nm.on_clicked(lambda e: nav_mouse(+1))
b_pd.on_clicked(lambda e: nav_day(-1));   b_nd.on_clicked(lambda e: nav_day(+1))

def on_search(text):
    t = text.strip()
    if not t: return
    hits = [m for m in mice if m == t] or [m for m in mice if m.startswith(t)] or [m for m in mice if t in m]
    if hits:
        state["mi"] = mice.index(hits[0]); state["di"] = 0; reload_rec()
    else:
        fig.suptitle(f"no mouse matching '{t}'", fontsize=10); fig.canvas.draw_idle()
tb.on_submit(on_search)

s_frame.on_changed(update)
s_trail.on_changed(lambda v: update(s_frame.val))
fig.canvas.mpl_connect("resize_event", lambda e: (fit_video_axes(), S.update(bg=None), update(int(s_frame.val))))

timer = fig.canvas.new_timer(interval=PLAY_INTERVAL)
def _tick():
    v = s_frame.val + int(s_speed.val)
    s_frame.set_val(s_frame.valmin if v > s_frame.valmax else v)
timer.add_callback(_tick)
playing = {"on": False}
def toggle_play(e):
    if playing["on"]: timer.stop(); b_play.label.set_text("Play")
    else: timer.start(); b_play.label.set_text("Pause")
    playing["on"] = not playing["on"]
b_play.on_clicked(toggle_play)

reload_rec()
try:
    mng = fig.canvas.manager
    try:    mng.window.showMaximized()
    except Exception:
        try:    mng.window.state("zoomed")
        except Exception: mng.window.wm_attributes("-zoomed", True)
except Exception:
    pass
plt.show()
```
