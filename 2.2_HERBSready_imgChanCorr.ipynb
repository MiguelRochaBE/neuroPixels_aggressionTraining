```python
import tifffile
from pathlib import Path
import numpy as np
import shutil
```

```python
root_DIR = Path(r"J:\project_trainingAggression\histologyData")

imgs = [f for f in root_DIR.rglob("*probeTrajectoryImages*\*.tif")]

for i in imgs:
    img = tifffile.imread(i)
    n_ch = np.shape(img)[0]

    if n_ch > 2:

        dest = i.parent / f"{i.stem}_2ch{i.suffix}" 

        out = img[[0, -1], :, :]     
        tifffile.imwrite(dest,
                         out,
                         imagej=True,
                         metadata={'axes': 'CYX'})
        
        outdir = i.parent / f"{str(n_ch)}_channelImages"
        outdir.mkdir(parents=True, exist_ok=True)
        shutil.move(str(i), str(outdir / i.name))
        
        print(f"Convert: {i}  (n_channels = {n_ch}) to 2 channels")
    
```
