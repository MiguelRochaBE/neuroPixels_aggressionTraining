```python
from pathlib import Path
import shutil
import re


# ============================================================
# PATHS
# ============================================================

rawVideos = Path(
    r"C:\Users\stagk\OneDrive\Desktop\trainingAggression_MR"
    r"\trainingSessions\RI_datasets\trainingSessions"
)

DLC_analyzed_videos = Path(
    r"C:\Users\stagk\OneDrive\Desktop\trainingAggression_MR"
    r"\trainingSessions\trainingSessions_topView_DLC_multiAnimal"
    r"\trainingSessions_topView_DLC_multiAnimal"
)

videosToAnalyze = Path(
    r"C:\Users\stagk\OneDrive\Desktop\trainingAggression_MR\trainingSessions\trainingSessions_topView_DLC_multiAnimal\trainingSessions_videosToAnalyze"
)


# ============================================================
# GET ALL DLC PATHS
# ============================================================

# Store every DLC path as lowercase text for matching
dlc_paths = [
    str(path).lower()
    for path in DLC_analyzed_videos.rglob("*")
]


# ============================================================
# SEARCH RAW VIDEOS
# ============================================================

copied = 0
already_analyzed = 0
could_not_parse = 0

# Find every video ending in "_topView_comp.avi"
for video in rawVideos.rglob("*_topView_comp.avi"):

    if not video.is_file():
        continue

    if "preTrain" in str(video):
        continue

    # Example:
    # mouse975830_Day04__topView_comp.avi

    filename = video.name

    # --------------------------------------------------------
    # Extract mouseID
    # --------------------------------------------------------
    mouse_match = re.search(r"(mouse\d+)", filename, re.IGNORECASE)

    # --------------------------------------------------------
    # Extract DayID
    # --------------------------------------------------------
    day_match = re.search(r"(Day\d+)", filename, re.IGNORECASE)

    if mouse_match is None or day_match is None:
        print(f"[COULD NOT PARSE] {video}")
        could_not_parse += 1
        continue

    mouseID = mouse_match.group(1)
    DayID = day_match.group(1)

    mouse_lower = mouseID.lower()
    day_lower = DayID.lower()

    # --------------------------------------------------------
    # Check DLC results
    #
    # A video counts as analyzed when BOTH mouseID and DayID
    # occur in the same DLC path.
    # --------------------------------------------------------
    is_analyzed = any(
        mouse_lower in dlc_path and day_lower in dlc_path
        for dlc_path in dlc_paths
    )

    if is_analyzed:
        print(f"[ANALYZED] {mouseID} / {DayID}")
        already_analyzed += 1
        continue

    # --------------------------------------------------------
    # Reproduce the raw folder structure relative to rawVideos
    #
    # Example:
    #
    # rawVideos/
    #   20250825_mouse975830_trainingSessions/
    #       Day04_/
    #           mouse975830_Day04__topView_comp.avi
    #
    # becomes:
    #
    # videosToAnalyze/
    #   20250825_mouse975830_trainingSessions/
    #       Day04_/
    #           mouse975830_Day04__topView_comp.avi
    # --------------------------------------------------------

    relative_path = video.relative_to(rawVideos)
    destination_file = videosToAnalyze / relative_path

    destination_file.parent.mkdir(parents=True, exist_ok=True)

    print(f"[COPY] {mouseID} / {DayID}")
    print(f"       {video}")
    print(f"    -> {destination_file}")

    shutil.copy2(video, destination_file)

    copied += 1


# ============================================================
# SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINISHED")
print("=" * 60)
print(f"Already analyzed : {already_analyzed}")
print(f"Copied           : {copied}")
print(f"Could not parse  : {could_not_parse}")
print(f"\nOutput folder:\n{videosToAnalyze}")
```
